```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('../netflix_titles.csv')

df.head()
```

```python
df = df.dropna(subset=['date_added'])
df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
df['year_added'] = df['date_added'].dt.year

def categorize_age(year):
    if year >= 2020:
        return 'New'
    elif year >= 2010:
        return 'Recent'
    else:
        return 'Classic'

df['age_category'] = df['release_year'].apply(categorize_age)

print(df.groupby('age_category').size())
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['release_year'], bins=30, color='blue')
plt.title('Distribution of Content Release Years')
plt.show()

added_per_year = df.groupby('year_added').size()
plt.figure(figsize=(10, 5))
added_per_year.plot(kind='line', marker='o', color='red')
plt.title('Netflix Titles Added Per Year')
plt.ylabel('Number of Titles')
plt.show()
```

```python
### Interpretation of Results
1. **Distribution:** The histogram shows that the majority of Netflix's library consists of content released after 2010, with a significant peak in very recent years.
2. **Trends:** The line chart demonstrates an exponential growth in content added starting around 2015, peaking in 2019 before a slight decline.
```
